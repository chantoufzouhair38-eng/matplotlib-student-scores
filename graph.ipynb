```python
import pandas as pd
import numpy as np

data = {
    'Student': ['Ali', 'Sara', 'Youssef', 'Lina', 'Omar',
                'Maya', 'Adam', 'Nour'],
    'Math': [85, 90, 72, 95, 60, 88, 78, 92],
    'Science': [80, 85, 75, 90, 65, 92, 82, 88],
    'English': [78, 92, 80, 88, 70, 85, 75, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

plt.bar(df['Student'], df['Math'])
plt.xlabel('Students')
plt.ylabel('Math Score')
plt.title('Students Math Scores')
plt.show()
```

```python
plt.plot(df["Student"],df["Math"],label="math")
plt.plot(df["Student"],df["Science"],label="science")
plt.legend()
plt.xlabel("student")
plt.ylabel("score")
plt.title("science vs math")
plt.show()
```

```python
import numpy as np
value_avg=[df["Math"].mean(),
df["Science"].mean(),
df["English"].mean()]
score_avg=np.array(value_avg)
cat=["math","scince","english"]
plt.pie(score_avg,labels=cat,autopct="%1.1f")
plt.title("avg score")
plt.show()
```

```python
plt.hist(df["Math"])
plt.xlabel("score math")
plt.ylabel("count")
plt.title("distribution of Math scores")
```

```python
plt.scatter(df["Student"],df["Math"],label="math")
plt.scatter(df["Student"],df["Science"],label="science")
plt.legend()
plt.title("Math scores with Science scores")
plt.xlabel("student")
plt.ylabel("score")
plt.show()
```

```python
plt.boxplot([df['Math'], df['Science'], df['English']],label=[1,2,2])

plt.title('Distribution of Students Scores')
plt.xlabel('Subjects')
plt.ylabel('Scores')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```
